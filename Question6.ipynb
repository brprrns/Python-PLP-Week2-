```python
def safe_divide(dividend, divisor):
    """Divide dividend by divisor.

    Args:
        dividend: The number to be divided.
        divisor: The number to divide by.

    Returns:
        The result of the division.

    Raises:
        ZeroDivisionError: If divisor is zero.
        TypeError: If dividend or divisor is not a number.
    """
    try:
        return dividend / divisor
    except ZeroDivisionError:
        raise ZeroDivisionError("Cannot divide by zero.")
    except TypeError:
        raise TypeError("Dividend and divisor must be numbers.")
```

**(c) Add type hints: (dividend: float, divisor: float) -> Optional[float].**

```python
from typing import Optional


def safe_divide(
    dividend: float,
    divisor: float
) -> Optional[float]:
    """Divide dividend by divisor.

    Args:
        dividend: The number to be divided.
        divisor: The number to divide by.

    Returns:
        The result of the division, or None if division fails.

    Raises:
        ZeroDivisionError: If divisor is zero.
        TypeError: If dividend or divisor is not a number.
    """
    try:
        return dividend / divisor
    except ZeroDivisionError:
        raise ZeroDivisionError("Cannot divide by zero.")
    except TypeError:
        raise TypeError("Dividend and divisor must be numbers.")
```

**(d) Write batch_divide(pairs: list[tuple]) calling safe_divide, skipping failures, logging 
warnings.**

```python
import logging
logging.basicConfig(level=logging.WARNING)

def batch_divide(pairs: list[tuple]) -> list[float]:
    """Divide multiple pairs and skip failed divisions.

    Args:
        pairs: A list of dividend and divisor pairs.

    Returns:
        A list containing successful division results.
    """
    results = []

    for dividend, divisor in pairs:
        try:
            result = safe_divide(dividend, divisor)
            results.append(result)
        except (ZeroDivisionError, TypeError) as error:
            logging.warning("Division failed: %s", error)

    return results
```

```python
pairs = [
    (10, 2),
    (6, 0),
    (9, 3),
    ("10", 2)
]

results = batch_divide(pairs)

print(results)
```
